# Food Delivery Time Prediction: End-to-End Notebook

**Goal:** predict `Time_taken_min` (total delivery time in minutes) using only information available **when the order is dispatched**.

**Key design decisions (explained in the project synopsis):**
1. `Average_Speed_kmph` is the speed measured *during* the trip, so it is **leakage** for a dispatch-time ETA. The primary model (Scenario A) excludes it. Scenario B includes it only as a diagnostic, to test whether it adds anything beyond dispatch-time information.
2. `Delivery_Distance_Category` is derived from `Road_Distance_km` (and 94.5% of rows are "Long"), so it is dropped. `Order_ID` is an identifier and is dropped.
3. The split is **chronological**: Jan-Aug train, Sep-Oct validation, Nov-Dec test. The test set is used **once**, at the end.
4. Every fitted step lives inside one scikit-learn `Pipeline`, so new data receives exactly the same transformations.
5. Feature engineering was tested one change at a time (Section 17). The final model uses the lean v2 feature set, a `total_expected_min` column and predictions clipped to the 180-minute cap.

**How to run:** place `Food_Delivery_Time_Prediction.csv` next to this notebook (or change `DATA_PATH`) and run all cells from top to bottom. All numbers you see come from your own run; nothing is hard-coded.

### Target analysis and leakage check

**What:** look at the target distribution, check whether the maximum value looks like a cap, and test how strongly `Average_Speed_kmph` gives away the answer.
**Why:** if a simple formula built from after-the-fact speed already reproduces the target, using speed would make our model look great but be useless at order time.

```python
print(df.shape, TARGET)
```

```python
y = df[TARGET]
print(y.describe().round(2).to_string())
print("Skewness:", round(y.skew(), 3))
n_max = int((y == y.max()).sum())
print(f"Rows exactly at the maximum ({y.max()} min): {n_max} ({n_max / len(y):.1%})  -> possible upper cap (censored values)")

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].hist(y, bins=50, color="#4C78A8"); ax[0].set_title("Delivery time distribution"); ax[0].set_xlabel("minutes")
ax[1].boxplot(y, vert=False); ax[1].set_title("Delivery time box plot"); ax[1].set_xlabel("minutes")
plt.tight_layout(); plt.show()
```

```python
tmp = df.copy()
tmp["formula_time"] = tmp["Road_Distance_km"] / tmp["Average_Speed_kmph"] * 60 + tmp["Preparation_Time_Min"]
print("Correlation of (distance / speed * 60 + prep time) with the target:",
      round(tmp[["formula_time", TARGET]].corr().iloc[0, 1], 3))
display((tmp[TARGET] - tmp["formula_time"]).describe().round(2).to_frame("actual minus formula (min)").T)
print("\nConclusion: speed + distance + prep time almost reproduce the target, so speed is excluded from the main model.")
```

## 4. Outliers

**What:** count values outside 1.5 x IQR for each numeric column. **Why:** decide whether to remove, cap or keep them.
**Decision:** keep them. Most are plausible (very experienced riders, long trips), and tree-based models are robust to them. Deleting real orders would make the test less honest.

```python
rows = []
for col in NUMERIC + ["Average_Speed_kmph", TARGET]:
    q1, q3 = df[col].quantile([0.25, 0.75]); iqr = q3 - q1
    n = int(((df[col] < q1 - 1.5 * iqr) | (df[col] > q3 + 1.5 * iqr)).sum())
    rows.append({"column": col, "outliers_IQR": n, "share_%": round(100 * n / len(df), 2)})
display(pd.DataFrame(rows).sort_values("outliers_IQR", ascending=False))
```

## 5. Time-based split

**What:** train on earlier months, tune on the next months, and test on the final months.
**Why:** a real ETA model is always used on *future* orders. A random split would let the model "see" the future and can overstate performance.
**Interview line:** "I split by date, and I touch the test months exactly once at the end."

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df = df.sort_values("Order_Date").reset_index(drop=True)

train = df[df["Order_Date"] < "2025-09-01"].copy()
valid = df[(df["Order_Date"] >= "2025-09-01") & (df["Order_Date"] < "2025-11-01")].copy()
test  = df[df["Order_Date"] >= "2025-11-01"].copy()
trainval = pd.concat([train, valid])

for name, part in [("train", train), ("valid", valid), ("test", test)]:
    print(f"{name:6s} rows={len(part):6d}  {part['Order_Date'].min().date()} -> {part['Order_Date'].max().date()}  mean target={part[TARGET].mean():.1f}")
```

## 6. Reusable code (`src/` package)

**What:** three small modules written to disk: feature engineering + pipeline, input validation, and prediction.
**Why:** the Streamlit app will import the *same* code, so training and serving can never drift apart. Putting the custom class in a module (not in the notebook) is also required for `joblib` to reload the saved pipeline later.

**Feature engineering inside the pipeline (`FeatureBuilder`)**
- month and day of month from the date
- hour as sine/cosine, and a rush-hour flag (8-10 and 18-21, from the EDA)
- same-zone flag (pickup = drop-off)
- ordinal encoding for traffic, restaurant load and priority
- preparation time per item
- `expected_travel_min` = distance / typical speed x 60, where the typical speed is the **median speed of that (vehicle, traffic, weather) group learned from training data only**. This is a safe replacement for the leaky individual-trip speed.

```python
%%writefile src/features.py
import numpy as np
import pandas as pd
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Columns a user must supply for a dispatch-time prediction
INPUT_COLUMNS = [
    "Order_Date", "Order_Hour", "Day_of_Week", "Is_Weekend", "Is_Festival", "Weather",
    "Pickup_Zone", "Dropoff_Zone", "Vehicle_Type", "Rider_Experience_Years", "Rider_Rating",
    "Restaurant_Rating", "Cuisine_Type", "Order_Items", "Restaurant_Load",
    "Preparation_Time_Min", "Road_Distance_km", "Traffic_Level", "Number_of_Signals",
    "Delivery_Priority",
]
NOMINAL_COLS = ["Day_of_Week", "Weather", "Pickup_Zone", "Dropoff_Zone", "Vehicle_Type", "Cuisine_Type"]
CATEGORICAL_INPUTS = NOMINAL_COLS + ["Restaurant_Load", "Traffic_Level", "Delivery_Priority"]
TRAFFIC_ORDER = {"Low": 0, "Moderate": 1, "High": 2, "Severe": 3}
LOAD_ORDER = {"Low": 0, "Medium": 1, "High": 2}
PRIORITY_ORDER = {"Normal": 0, "Priority": 1, "VIP": 2}
GROUP_COLS = ["Vehicle_Type", "Traffic_Level", "Weather"]


class FeatureBuilder(BaseEstimator, TransformerMixin):
    # Builds model-ready columns from raw order columns. fit() learns typical speeds per
    # (vehicle, traffic, weather) group from TRAINING data only. transform() never needs the
    # individual trip's speed unless use_actual_speed=True (Scenario B, diagnostic only).

    def __init__(self, use_prep_time=True, use_actual_speed=False):
        self.use_prep_time = use_prep_time
        self.use_actual_speed = use_actual_speed

    def fit(self, X, y=None):
        X = pd.DataFrame(X)
        needed = GROUP_COLS + ["Average_Speed_kmph"]
        missing = [c for c in needed if c not in X.columns]
        if missing:
            raise ValueError(f"Training data must contain {missing} to learn typical speeds.")
        tmp = X[GROUP_COLS].copy()
        tmp["speed"] = pd.to_numeric(X["Average_Speed_kmph"], errors="coerce")
        self.global_speed_ = float(tmp["speed"].median())
        self.group_speed_ = tmp.groupby(GROUP_COLS)["speed"].median().rename("group_speed").reset_index()
        return self

    def transform(self, X):
        X = pd.DataFrame(X).copy()
        missing = [c for c in INPUT_COLUMNS if c not in X.columns]
        if missing:
            raise ValueError(f"Missing required columns: {missing}")
        num = lambda c: pd.to_numeric(X[c], errors="coerce")
        out = pd.DataFrame(index=X.index)

        hour = num("Order_Hour")
        out["Order_Hour"] = hour
        out["hour_sin"] = np.sin(2 * np.pi * hour / 24)
        out["hour_cos"] = np.cos(2 * np.pi * hour / 24)
        out["is_rush_hour"] = (hour.between(8, 10) | hour.between(18, 21)).astype(float)

        date = pd.to_datetime(X["Order_Date"], errors="coerce")
        out["month"] = date.dt.month
        out["day_of_month"] = date.dt.day

        for c in ["Is_Weekend", "Is_Festival", "Rider_Experience_Years", "Rider_Rating",
                  "Restaurant_Rating", "Order_Items", "Road_Distance_km", "Number_of_Signals",
                  "Preparation_Time_Min"]:
            out[c] = num(c)

        out["same_zone"] = (X["Pickup_Zone"] == X["Dropoff_Zone"]).astype(float)
        out["traffic_level_ord"] = X["Traffic_Level"].map(TRAFFIC_ORDER)
        out["restaurant_load_ord"] = X["Restaurant_Load"].map(LOAD_ORDER)
        out["priority_ord"] = X["Delivery_Priority"].map(PRIORITY_ORDER)
        out["prep_per_item"] = out["Preparation_Time_Min"] / out["Order_Items"].replace(0, np.nan)

        grp = X[GROUP_COLS].merge(self.group_speed_, on=GROUP_COLS, how="left")["group_speed"].to_numpy()
        grp = np.where(np.isnan(grp), self.global_speed_, grp)
        out["expected_travel_min"] = out["Road_Distance_km"].to_numpy() / grp * 60

        if self.use_actual_speed:
            out["Average_Speed_kmph"] = num("Average_Speed_kmph")

        for c in NOMINAL_COLS:
            out[c] = X[c].astype("object")
        return out


def numeric_feature_names(use_prep_time=True, use_actual_speed=False):
    cols = ["Order_Hour", "hour_sin", "hour_cos", "is_rush_hour", "Is_Weekend", "Is_Festival",
            "month", "day_of_month", "same_zone", "Rider_Experience_Years", "Rider_Rating",
            "Restaurant_Rating", "Order_Items", "Road_Distance_km", "Number_of_Signals",
            "traffic_level_ord", "restaurant_load_ord", "priority_ord", "expected_travel_min"]
    if use_prep_time:
        cols += ["Preparation_Time_Min", "prep_per_item"]
    if use_actual_speed:
        cols += ["Average_Speed_kmph"]
    return cols


def build_pipeline(model, scale=False, use_prep_time=True, use_actual_speed=False):
    num_steps = [("imputer", SimpleImputer(strategy="median"))]
    if scale:
        num_steps.append(("scaler", StandardScaler()))
    preprocess = ColumnTransformer(
        [("num", Pipeline(num_steps), numeric_feature_names(use_prep_time, use_actual_speed)),
         ("cat", Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),
                           ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))]),
          NOMINAL_COLS)],
        remainder="drop",
    )
    return Pipeline([
        ("features", FeatureBuilder(use_prep_time=use_prep_time, use_actual_speed=use_actual_speed)),
        ("preprocess", preprocess),
        ("model", model),
    ])
```

**Improved feature set (v2), built on top of the code above.** `src/features_v2.py` keeps `FeatureBuilder` as it is and adds three things, all chosen by the experiments in **Section 17** (read that section for the evidence):
- a lean column list: columns that made no difference in the tests are removed (weekday/weekend flags, same-zone flag, hour sin/cos, month, day of month, ratings, restaurant load); `Delivery_Priority` is kept because removing it made the model worse
- one new column, `total_expected_min` = expected travel time + preparation time
- predictions are clipped to the range of the training target (10 to 180 min), so the model can never predict beyond the 180-minute cap

Everything from Section 7 onwards uses this v2 pipeline. The original v1 pipeline is kept only for the v1 vs v2 comparison in Section 17.

```python
%%writefile src/features_v2.py
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from src.features import FeatureBuilder

# Lean feature set chosen from the Section 17 experiments (validation months only)
V2_NUMERIC = ["is_rush_hour", "Is_Festival", "Rider_Experience_Years", "Order_Items",
              "Road_Distance_km", "Number_of_Signals", "traffic_level_ord", "priority_ord",
              "expected_travel_min", "Preparation_Time_Min", "prep_per_item", "total_expected_min"]
V2_PREP_COLS = ["Preparation_Time_Min", "prep_per_item", "total_expected_min"]
V2_NOMINAL = ["Weather", "Vehicle_Type", "Cuisine_Type", "Pickup_Zone", "Dropoff_Zone"]


class FeatureBuilderV2(FeatureBuilder):
    # Same as FeatureBuilder plus total_expected_min = expected travel time + preparation time.
    def transform(self, X):
        out = super().transform(X)
        out["total_expected_min"] = out["expected_travel_min"] + out["Preparation_Time_Min"]
        return out


class ClippedPipeline(Pipeline):
    # A normal Pipeline whose predictions are clipped to the range of the TRAINING target
    # (read inside fit, so nothing is hard-coded). Parameter names (model__..., preprocess__...)
    # stay the same as in a plain Pipeline, so tuning and cloning work unchanged.
    def fit(self, X, y=None, **kwargs):
        super().fit(X, y, **kwargs)
        self.low_, self.high_ = float(np.min(y)), float(np.max(y))
        return self

    def predict(self, X, **kwargs):
        return np.clip(super().predict(X, **kwargs), self.low_, self.high_)


def numeric_feature_names_v2(use_prep_time=True, use_actual_speed=False):
    cols = [c for c in V2_NUMERIC if use_prep_time or c not in V2_PREP_COLS]
    return cols + (["Average_Speed_kmph"] if use_actual_speed else [])


def build_pipeline_v2(model, scale=False, use_prep_time=True, use_actual_speed=False):
    num_steps = [("imputer", SimpleImputer(strategy="median"))]
    if scale:
        num_steps.append(("scaler", StandardScaler()))
    preprocess = ColumnTransformer(
        [("num", Pipeline(num_steps), numeric_feature_names_v2(use_prep_time, use_actual_speed)),
         ("cat", Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),
                           ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))]), V2_NOMINAL)],
        remainder="drop")
    return ClippedPipeline([
        ("features", FeatureBuilderV2(use_prep_time=use_prep_time, use_actual_speed=use_actual_speed)),
        ("preprocess", preprocess),
        ("model", model),
    ])
```

```python
%%writefile src/validation.py
import numpy as np
import pandas as pd

# Hard rules: a value outside these limits is impossible, so the row is REJECTED.
HARD_LIMITS = {
    "Order_Hour": (0, 23), "Is_Weekend": (0, 1), "Is_Festival": (0, 1),
    "Rider_Experience_Years": (0, 60), "Rider_Rating": (1, 5), "Restaurant_Rating": (1, 5),
    "Order_Items": (1, None), "Preparation_Time_Min": (0, None),
    "Road_Distance_km": (0.001, None), "Number_of_Signals": (0, None),
}


def validate_orders(df, meta):
    # Returns a DataFrame with 'status' (Valid / Low confidence / Rejected) and 'reasons' per row.
    # Raises ValueError if required columns are missing (the whole file is unusable).
    df = df.reset_index(drop=True)
    missing = [c for c in meta["required_columns"] if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns: {missing}")

    n = len(df)
    reject = [[] for _ in range(n)]
    low = [[] for _ in range(n)]

    def add(store, mask, msg):
        for i in np.where(np.asarray(mask))[0]:
            store[i].append(msg)

    # Categorical columns must contain a category seen in training
    for col, allowed in meta["categories"].items():
        add(reject, ~df[col].isin(allowed), f"{col}: missing or unknown value")

    # Numeric columns must be numbers, inside hard limits; outside training range = low confidence
    for col, (lo_hard, hi_hard) in HARD_LIMITS.items():
        num = pd.to_numeric(df[col], errors="coerce")
        add(reject, num.isna(), f"{col}: missing or not a number")
        if lo_hard is not None:
            add(reject, num.notna() & (num < lo_hard), f"{col}: below allowed minimum {lo_hard}")
        if hi_hard is not None:
            add(reject, num.notna() & (num > hi_hard), f"{col}: above allowed maximum {hi_hard}")
        if col in meta["numeric_ranges"]:
            lo, hi = meta["numeric_ranges"][col]
            add(low, num.notna() & ((num < lo) | (num > hi)), f"{col}: outside training range [{lo}, {hi}]")

    # Date checks
    date = pd.to_datetime(df["Order_Date"], errors="coerce")
    add(reject, date.isna(), "Order_Date: missing or not a valid date")
    add(low, date.notna() & (date.dt.day_name() != df["Day_of_Week"]), "Day_of_Week does not match Order_Date")
    weekend = pd.to_numeric(df["Is_Weekend"], errors="coerce")
    add(low, df["Day_of_Week"].isin(["Saturday", "Sunday"]) != (weekend == 1), "Is_Weekend does not match Day_of_Week")

    status = ["Rejected" if reject[i] else ("Low confidence" if low[i] else "Valid") for i in range(n)]
    reasons = ["; ".join(reject[i] + low[i]) for i in range(n)]
    return pd.DataFrame({"status": status, "reasons": reasons})
```

```python
%%writefile src/predict.py
import numpy as np
import pandas as pd
from .validation import validate_orders


def predict_orders(df, bundle):
    # Validate, then predict only for non-rejected rows. Never labels a prediction right or wrong:
    # that is only possible when a true value exists (see compare_with_actual).
    meta = bundle["meta"]
    df = df.reset_index(drop=True)
    checks = validate_orders(df, meta)
    result = pd.concat([df, checks], axis=1)
    for c in ["predicted_min", "range_low_min", "range_high_min"]:
        result[c] = np.nan

    ok = (checks["status"] != "Rejected").to_numpy()
    if ok.any():
        X = df.loc[ok, meta["required_columns"]]
        pred = bundle["model"].predict(X)
        lo = np.minimum(bundle["q_low"].predict(X), pred)
        hi = np.maximum(bundle["q_high"].predict(X), pred)
        t_lo, t_hi = meta["target_range"]
        result.loc[ok, "predicted_min"] = np.clip(pred, t_lo, t_hi).round(1)
        result.loc[ok, "range_low_min"] = np.clip(lo, t_lo, t_hi).round(1)
        result.loc[ok, "range_high_min"] = np.clip(hi, t_lo, t_hi).round(1)

        wide = ok & ((result["range_high_min"] - result["range_low_min"]) > meta["wide_interval_threshold"]).to_numpy()
        for i in np.where(wide & (result["status"] == "Valid").to_numpy())[0]:
            result.loc[i, "status"] = "Low confidence"
            result.loc[i, "reasons"] = "prediction range is unusually wide for this kind of order"
    return result


def compare_with_actual(result, tolerance=10.0, target="Time_taken_min"):
    # Only valid when the true value is present in the uploaded data.
    if target not in result.columns:
        raise ValueError(f"'{target}' is not in the data, so correctness cannot be judged.")
    out = result.copy()
    out["abs_error_min"] = (out["predicted_min"] - out[target]).abs().round(1)
    out["result"] = np.where(out["predicted_min"].isna(), "Not predicted",
                             np.where(out["abs_error_min"] <= tolerance, f"Correct (within {tolerance:g} min)",
                                      f"Wrong (off by more than {tolerance:g} min)"))
    return out
```

```python
from src.features import INPUT_COLUMNS, CATEGORICAL_INPUTS, build_pipeline as build_pipeline_v1   # original features (for Section 17 only)
from src.features_v2 import build_pipeline_v2 as build_pipeline                                  # improved features used from here on
from src.validation import validate_orders
from src.predict import predict_orders, compare_with_actual

def X_of(d):
    # Dispatch-time inputs. Average_Speed_kmph is passed ONLY so that FeatureBuilder.fit can learn
    # typical speeds from training rows; transform() ignores it in Scenario A.
    return d[INPUT_COLUMNS + ["Average_Speed_kmph"]]

def regression_report(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float); y_pred = np.asarray(y_pred, dtype=float)
    err = np.abs(y_true - y_pred)
    return {"MAE": mean_absolute_error(y_true, y_pred),
            "RMSE": float(np.sqrt(mean_squared_error(y_true, y_pred))),
            "R2": r2_score(y_true, y_pred),
            "Within_10min_%": float((err <= 10).mean() * 100),
            "P90_abs_error": float(np.percentile(err, 90))}

def fit_eval(model, fit_df, eval_df, scale=False, use_prep=True, use_speed=False):
    pipe = build_pipeline(model, scale=scale, use_prep_time=use_prep, use_actual_speed=use_speed)
    pipe.fit(X_of(fit_df), fit_df[TARGET])
    return pipe, regression_report(eval_df[TARGET], pipe.predict(X_of(eval_df)))

print("Required input columns:", len(INPUT_COLUMNS))
```

## 7. Baseline and model comparison (Scenario A: dispatch-time features)

**What:** train a mean-prediction baseline, Ridge regression, a Decision Tree, a Random Forest and Gradient Boosting. Train on the training months, score on the **validation** months.
**Why:** a model is only useful if it beats a trivial baseline; comparing simple and complex models shows whether complexity is justified.
**Metric to pick the winner:** validation MAE (average error in minutes).

```python
models = {
    "Mean baseline":          (DummyRegressor(strategy="mean"), False),
    "Ridge":                  (Ridge(alpha=1.0), True),
    "Decision Tree":          (DecisionTreeRegressor(max_depth=8, min_samples_leaf=20, random_state=RANDOM_STATE), False),
    "Random Forest":          (RandomForestRegressor(n_estimators=200, min_samples_leaf=3, n_jobs=-1, random_state=RANDOM_STATE), False),
    "Hist Gradient Boosting": (HistGradientBoostingRegressor(random_state=RANDOM_STATE), False),
}

fitted, rows = {}, []
for name, (model, scale) in models.items():
    pipe, rep = fit_eval(model, train, valid, scale=scale)
    fitted[name] = pipe
    rows.append({"model": name, **rep})

comparison = pd.DataFrame(rows).sort_values("MAE").reset_index(drop=True)
display(comparison.round(3))
```

### 7.1 Ablation and leakage demonstration

- **Without preparation time:** the recorded prep time is the *actual* time the kitchen took, which a platform only knows exactly after the food is ready. We assume an *estimate* is available at dispatch; this test shows how much the model depends on that assumption.
- **Scenario B (with `Average_Speed_kmph`):** diagnostic only. If this looks far better than Scenario A, that gap is the leakage we avoided.

```python
_, rep_noprep = fit_eval(HistGradientBoostingRegressor(random_state=RANDOM_STATE), train, valid, use_prep=False)
_, rep_speed  = fit_eval(HistGradientBoostingRegressor(random_state=RANDOM_STATE), train, valid, use_speed=True)

ablation = pd.DataFrame([
    {"setup": "Scenario A (primary): dispatch-time features", **comparison.set_index("model").loc["Hist Gradient Boosting"].to_dict()},
    {"setup": "Scenario A without preparation time", **rep_noprep},
    {"setup": "Scenario B (DIAGNOSTIC ONLY): adds actual trip speed", **rep_speed},
])
display(ablation.round(3))
gap = ablation.loc[0, "MAE"] - ablation.loc[2, "MAE"]
print(f"MAE gap (Scenario A minus Scenario B): {gap:.3f} min")
if gap > 1:
    print("Scenario B is clearly better: this gap is the leakage effect. It is NOT deployable, because speed is only known after delivery.")
else:
    print("The gap is tiny: in this dataset trip speed adds almost nothing beyond vehicle, traffic and weather, which Scenario A already uses.")
    print("Excluding speed costs nothing here and keeps the model usable at dispatch time. Scenario B is still not deployable.")
```

**What:** compare a shuffled 5-fold CV score with the chronological validation score for the same model.
**Why:** if the two differ noticeably, the data changes over time (drift) and a random split would be misleading. If they are similar, we learn the pattern is stable across months.

```python
cv_rows = []
for name in ["Ridge", "Hist Gradient Boosting"]:
    model, scale = models[name]
    pipe = build_pipeline(clone(model), scale=scale)
    scores = -cross_val_score(pipe, X_of(trainval), trainval[TARGET], cv=KFold(5, shuffle=True, random_state=RANDOM_STATE),
                              scoring="neg_mean_absolute_error", n_jobs=1)
    cv_rows.append({"model": name,
                    "random 5-fold MAE (mean)": scores.mean(),
                    "random 5-fold MAE (std)": scores.std(),
                    "time-based validation MAE": comparison.set_index("model").loc[name, "MAE"]})
display(pd.DataFrame(cv_rows).round(3))
```

**What:** randomized search over Gradient Boosting settings, using `TimeSeriesSplit` inside the **training** months only.
**Why:** tuning on time-ordered folds respects the order of time and never touches validation or test data.

## 11. Prediction intervals

**What:** two extra models predict the 10th and 90th percentile of delivery time, giving a likely range (nominally 80%) around each prediction.
**Why:** an ETA of "70 min" and "70 min, likely 55-95" are very different promises. The range also lets the app flag unreliable orders.
The "unusually wide range" threshold is learned from validation-period ranges (models fitted on training months only), then the quantile models are refitted on train + validation.

```python
if final_name.startswith("Hist Gradient Boosting (tuned)"):
    q_params = {k: v for k, v in search.best_params_.items() if k != "model__loss"}
else:
    q_params = {}

def make_quantile_model(q):
    pipe = build_pipeline(HistGradientBoostingRegressor(loss="quantile", quantile=q, random_state=RANDOM_STATE))
    return pipe.set_params(**q_params)

# 1) widths on validation months from train-only models -> threshold for "unusually wide"
q10_tr = make_quantile_model(0.10).fit(X_of(train), train[TARGET])
q90_tr = make_quantile_model(0.90).fit(X_of(train), train[TARGET])
val_pred = fitted[final_name].predict(X_of(valid))
val_lo = np.minimum(q10_tr.predict(X_of(valid)), val_pred)
val_hi = np.maximum(q90_tr.predict(X_of(valid)), val_pred)
wide_threshold = float(np.percentile(val_hi - val_lo, 90))

# 2) refit on train + validation and check coverage on the untouched test months
q_low = make_quantile_model(0.10).fit(X_of(trainval), trainval[TARGET])
q_high = make_quantile_model(0.90).fit(X_of(trainval), trainval[TARGET])
lo = np.minimum(q_low.predict(X_of(test)), test_pred)
hi = np.maximum(q_high.predict(X_of(test)), test_pred)
coverage = float(((test[TARGET].values >= lo) & (test[TARGET].values <= hi)).mean())
print(f"Nominal coverage of the 10%-90% range: 80%  |  observed on test: {coverage:.1%}")
print(f"Average range width on test: {np.mean(hi - lo):.1f} min  |  'unusually wide' threshold: {wide_threshold:.1f} min")
```

## 16. Independent testing system (what an interviewer would do)

The interviewer can supply new orders in the same format. The system does this, in order:
1. **Validate** (missing columns stop the whole file; bad rows are marked *Rejected*; unusual-but-possible rows are marked *Low confidence*).
2. **Preprocess** using the saved pipeline (identical to training).
3. **Predict** with a likely range.
4. **Compare with actual values only if a true `Time_taken_min` column is present.**

Statuses mean: **Valid** = normal prediction; **Low confidence** = prediction given but treat with care; **Rejected** = no prediction because the input is impossible or incomplete. We never call a prediction "correct" or "wrong" without a true value.

```python
def make_case(label, date="2025-12-10", **overrides):
    base = {"Order_Date": date, "Order_Hour": 15, "Is_Festival": 0, "Weather": "Clear",
            "Pickup_Zone": "Residential", "Dropoff_Zone": "Commercial", "Vehicle_Type": "Bike",
            "Rider_Experience_Years": 4.0, "Rider_Rating": 4.0, "Restaurant_Rating": 4.2,
            "Cuisine_Type": "North Indian", "Order_Items": 2, "Restaurant_Load": "Low",
            "Preparation_Time_Min": 20, "Road_Distance_km": 8.0, "Traffic_Level": "Low",
            "Number_of_Signals": 6, "Delivery_Priority": "Normal"}
    base.update(overrides)
    d = pd.Timestamp(base["Order_Date"])
    base.setdefault("Day_of_Week", d.day_name())
    base.setdefault("Is_Weekend", int(d.dayofweek >= 5))
    base["Case"] = label
    return base

cases = pd.DataFrame([
    make_case("1 Normal: bike, clear, low traffic, 8 km"),
    make_case("2 Normal rush hour: scooter, rain, high traffic, 15 km", Order_Hour=19, Weather="Rain",
              Traffic_Level="High", Vehicle_Type="Scooter", Road_Distance_km=15, Restaurant_Load="High", Preparation_Time_Min=28, Number_of_Signals=12),
    make_case("3 Edge: very short trip (0.08 km)", Road_Distance_km=0.08, Number_of_Signals=2),
    make_case("4 Edge: very long trip (70 km)", Road_Distance_km=70, Number_of_Signals=18),
    make_case("5 Unusual: bicycle, storm, severe traffic, festival, 45 km", Vehicle_Type="Bicycle", Weather="Storm",
              Traffic_Level="Severe", Is_Festival=1, Road_Distance_km=45, Order_Hour=20, Number_of_Signals=16),
    make_case("6 Outside training range: 120 km", Road_Distance_km=120),
    make_case("7 Invalid: unknown vehicle 'Helicopter'", Vehicle_Type="Helicopter"),
    make_case("8 Invalid: hour = 27", Order_Hour=27),
    make_case("9 Invalid: negative distance", Road_Distance_km=-5),
    make_case("10 Invalid: text in a numeric field", Preparation_Time_Min="abc"),
])

out = predict_orders(cases.drop(columns="Case"), reloaded)
out.insert(0, "Case", cases["Case"].values)
pd.set_option("display.max_colwidth", 90)
display(out[["Case", "status", "predicted_min", "range_low_min", "range_high_min", "reasons"]])
```

### Predict a whole new file

Use this helper with any CSV/Excel file that has the required columns. If the file also contains `Time_taken_min`, the output includes error and Correct/Wrong columns; otherwise it only reports Valid / Low confidence / Rejected.

## 17. Feature engineering study: how the v2 features were chosen

**What:** a controlled test of every feature idea we listed: columns to remove, columns to add, and changes to how the target is predicted.
**Why:** a feature is only worth keeping if the model does better *without a doubt* on data it has not seen. So every idea is tested by itself against the current model, on the **validation** months only (the test months are not touched here).

**How the test works**
- Start from the features `FeatureBuilder` already makes (Section 6), then add the candidate columns on top.
- Each experiment changes **one thing**. The model is always the same (Hist Gradient Boosting, default settings).
- Every experiment is run with 3 random seeds and the MAE is averaged. The seeds give a *noise level* (about 0.005 min). A change smaller than **0.01 min** is treated as "no real change", because we cannot tell it apart from luck.
- Besides the overall MAE we also report the error on the rows where the delivery time is exactly 180 min (the cap seen in Section 2) and on all other rows.

**Where this fits:** the winning features (v2) are defined in Section 6 and already used in Sections 7 to 15. This section is the evidence behind them, plus the v1 vs v2 comparison. The notebook is written in this order so that the main flow reads top to bottom as one pipeline.

```python
from src.features import FeatureBuilder, NOMINAL_COLS, numeric_feature_names

fb = FeatureBuilder().fit(X_of(train))            # speeds learned from training rows only

def candidate_frame(d):
    """The pipeline's own features + extra candidate columns (all known at dispatch time)."""
    F = fb.transform(X_of(d))
    F["total_expected_min"] = F["expected_travel_min"] + F["Preparation_Time_Min"]       # travel + kitchen time
    F["log_distance"]       = np.log1p(F["Road_Distance_km"])
    F["bad_weather"]        = F["Weather"].isin(["Rain", "Storm", "Fog"]).astype(float)
    F["weather_ord"]        = F["Weather"].map({"Clear": 0, "Cloudy": 0.5, "Fog": 1, "Rain": 2, "Storm": 3})
    F["stress_score"]       = F["weather_ord"] + F["traffic_level_ord"]                   # weather + traffic combined
    F["dist_x_stress"]      = F["Road_Distance_km"] * (1 + F["stress_score"])
    F["signals_per_km"]     = F["Number_of_Signals"] / F["Road_Distance_km"].clip(lower=0.5)
    F["signals_x_stress"]   = F["Number_of_Signals"] * (1 + F["stress_score"])
    F["signals_x_traffic"]  = F["Number_of_Signals"] * (1 + F["traffic_level_ord"])
    F["prep_x_load"]        = F["Preparation_Time_Min"] * (1 + F["restaurant_load_ord"])
    F["exp_bin"]            = pd.cut(F["Rider_Experience_Years"], [0, 1, 3, 5, 100], labels=False)
    F["exp_x_dist"]         = F["Rider_Experience_Years"] * F["Road_Distance_km"]
    F["rush_x_weekend"]     = F["is_rush_hour"] * F["Is_Weekend"]
    return F

F_tr, F_va = candidate_frame(train), candidate_frame(valid)
y_tr, y_va = train[TARGET].to_numpy(float), valid[TARGET].to_numpy(float)
at_cap = (y_va == y_tr.max())                      # rows sitting exactly on the 180-minute cap

def evaluate(num, nom=NOMINAL_COLS, target="raw", clip=False, seeds=(0, 1, 2)):
    both = pd.concat([F_tr, F_va])
    M = both[num].copy()
    for c in nom:
        M = M.join(pd.get_dummies(both[c], prefix=c).astype(float))
    Mtr, Mva = M.iloc[:len(F_tr)], M.iloc[len(F_tr):]
    med = Mtr.median(); Mtr, Mva = Mtr.fillna(med), Mva.fillna(med)
    off_tr = F_tr["total_expected_min"].to_numpy() if target == "resid" else 0.0   # predict only the leftover
    off_va = F_va["total_expected_min"].to_numpy() if target == "resid" else 0.0
    maes, first = [], None
    for s in seeds:
        m = HistGradientBoostingRegressor(random_state=s)
        if target == "log":
            m.fit(Mtr, np.log(y_tr)); p = np.exp(m.predict(Mva))
        else:
            m.fit(Mtr, y_tr - off_tr); p = m.predict(Mva) + off_va
        if clip:
            p = np.clip(p, y_tr.min(), y_tr.max())
        maes.append(np.abs(y_va - p).mean())
        first = p if first is None else first
    err = np.abs(y_va - first)
    return {"MAE": np.mean(maes), "seed_std": np.std(maes),
            "MAE_on_cap_rows": err[at_cap].mean(), "MAE_other_rows": err[~at_cap].mean()}

BASE = numeric_feature_names()                     # exactly what the current model uses
without = lambda *cols: [c for c in BASE if c not in cols]
nom_without = lambda *cols: [c for c in NOMINAL_COLS if c not in cols]
TE = ["total_expected_min"]

experiments = [
    # (group, name, numeric cols, nominal cols, options)
    ("0 baseline", "Current model",                                BASE, NOMINAL_COLS, {}),
    ("1 drop",     "Drop Day_of_Week (keep Is_Weekend)",           BASE, nom_without("Day_of_Week"), {}),
    ("1 drop",     "Drop Is_Weekend (keep Day_of_Week)",           without("Is_Weekend"), NOMINAL_COLS, {}),
    ("1 drop",     "Drop same_zone flag",                          without("same_zone"), NOMINAL_COLS, {}),
    ("1 drop",     "Drop same_zone + Pickup/Dropoff zones",        without("same_zone"), nom_without("Pickup_Zone", "Dropoff_Zone"), {}),
    ("1 drop",     "Drop hour sin/cos",                            without("hour_sin", "hour_cos"), NOMINAL_COLS, {}),
    ("1 drop",     "Drop Order_Hour + sin/cos (keep rush flag)",   without("Order_Hour", "hour_sin", "hour_cos"), NOMINAL_COLS, {}),
    ("1 drop",     "Drop day_of_month",                            without("day_of_month"), NOMINAL_COLS, {}),
    ("1 drop",     "Drop month + day_of_month",                    without("month", "day_of_month"), NOMINAL_COLS, {}),
    ("1 drop",     "Drop Rider_Rating + Restaurant_Rating",        without("Rider_Rating", "Restaurant_Rating"), NOMINAL_COLS, {}),
    ("1 drop",     "Drop restaurant_load_ord",                     without("restaurant_load_ord"), NOMINAL_COLS, {}),
    ("1 drop",     "Drop priority_ord (Delivery_Priority)",        without("priority_ord"), NOMINAL_COLS, {}),
    ("1 drop",     "Drop Cuisine_Type",                            BASE, nom_without("Cuisine_Type"), {}),
    ("1 drop",     "Drop Order_Items + prep_per_item",             without("Order_Items", "prep_per_item"), NOMINAL_COLS, {}),
    ("2 add",      "Add log(distance)",                            BASE + ["log_distance"], NOMINAL_COLS, {}),
    ("2 add",      "Add bad_weather flag",                         BASE + ["bad_weather"], NOMINAL_COLS, {}),
    ("2 add",      "Add weather+traffic stress score",             BASE + ["weather_ord", "stress_score"], NOMINAL_COLS, {}),
    ("2 add",      "Add distance x stress",                        BASE + ["weather_ord", "stress_score", "dist_x_stress"], NOMINAL_COLS, {}),
    ("2 add",      "Add signals per km",                           BASE + ["signals_per_km"], NOMINAL_COLS, {}),
    ("2 add",      "Add signals x traffic",                        BASE + ["signals_x_traffic"], NOMINAL_COLS, {}),
    ("2 add",      "Add prep x restaurant load",                   BASE + ["prep_x_load"], NOMINAL_COLS, {}),
    ("2 add",      "Add rider experience bins",                    BASE + ["exp_bin"], NOMINAL_COLS, {}),
    ("2 add",      "Add experience x distance",                    BASE + ["exp_x_dist"], NOMINAL_COLS, {}),
    ("2 add",      "Add rush x weekend",                           BASE + ["rush_x_weekend"], NOMINAL_COLS, {}),
    ("2 add",      "Add total_expected_min (travel + prep)",       BASE + TE, NOMINAL_COLS, {}),
    ("3 target",   "Predict residual (actual - total_expected)",   BASE + TE, NOMINAL_COLS, {"target": "resid"}),
    ("3 target",   "Predict log(time)",                            BASE, NOMINAL_COLS, {"target": "log"}),
    ("3 target",   "Clip predictions to the training range",       BASE, NOMINAL_COLS, {"clip": True}),
    ("3 target",   "Residual target + clip",                       BASE + TE, NOMINAL_COLS, {"target": "resid", "clip": True}),
]

rows = []
for group, name, num, nom, opts in experiments:
    r = evaluate(num, nom, **opts); rows.append({"group": group, "experiment": name, **r})
exp = pd.DataFrame(rows)
base_mae = exp.loc[0, "MAE"]
exp["change_vs_baseline"] = exp["MAE"] - base_mae
NOISE = 0.01
exp["verdict"] = np.where(exp["change_vs_baseline"] <= -NOISE, "HELPS",
                  np.where(exp["change_vs_baseline"] >= NOISE, "HURTS", "no real change"))
display(exp.drop(columns="group").round(3))
exp.to_csv("reports/feature_experiments_validation.csv", index=False)
```

```python
plot = exp.iloc[1:].sort_values("change_vs_baseline")
colors = plot["verdict"].map({"HELPS": "#59A14F", "HURTS": "#E15759", "no real change": "#BAB0AC"})
fig, ax = plt.subplots(figsize=(9, 9))
ax.barh(plot["experiment"], plot["change_vs_baseline"], color=colors)
ax.axvspan(-NOISE, NOISE, color="grey", alpha=0.15, label=f"noise band (+/-{NOISE} min)")
ax.axvline(0, color="black", lw=0.8)
ax.set_xlabel("Change in validation MAE vs current model (minutes; negative = better)")
ax.set_title("Feature experiments, one change at a time"); ax.legend(loc="lower right"); plt.tight_layout(); plt.show()
```

### 17.1 What the experiments showed

**Columns to drop.** Removing any one of the weekday / weekend / zone / hour-encoding / date / rating / load / cuisine columns changes MAE by less than the noise band. These columns are not *hurting*, but they are not *helping* either: the model gets the same accuracy without them. A smaller feature set is easier to explain, cheaper to collect at dispatch time, and less likely to break when the data changes. So we drop them in the lean model (Section 6).

**Keep `Delivery_Priority`.** It is the one drop candidate that clearly hurts (MAE gets worse when removed). It is a small effect (VIP orders finish about 1.5 min faster than Normal orders, even after accounting for distance, traffic and weather) but it is real.

**Added columns that did not help.** `log(distance)`, `bad_weather`, the stress score, distance x stress, signals per km, signals x traffic, prep x load, experience bins and rush x weekend all land inside the noise band, and some are slightly worse. The reason: gradient boosting already finds these interactions by itself. Spelling them out only helps *linear* models (Ridge), which are far behind anyway.

**Added column that did help: `total_expected_min` = expected travel time + preparation time.** The model has to *discover* that the total time is roughly "travel + kitchen". Handing it that sum directly gives it one strong column to split on and lowers the error clearly (about 0.065 min), on both the capped rows and the normal rows.

**Target tricks.** Clipping predictions to the range seen in training (10 to 180 min) gives a small extra gain: the real data never goes above the 180 cap, so a prediction of 187 can only be wrong. Predicting the *residual* (actual minus `total_expected_min`) is better on normal rows (2.155 vs 2.240) but much worse on the capped rows, because a capped time is not "expected + noise"; it is simply cut off at 180. Even with a clip added, the residual target (2.159) is behind just adding `total_expected_min` as a feature and clipping (about 2.125, Section 6). A log target is worse overall. So we keep the raw target.

```python
def compare_v1_v2(fit_df, eval_df, seeds=(0, 1, 2)):
    builders = {"v1 (original features)": build_pipeline_v1,
                "v2 (lean + total_expected + clip)": build_pipeline}
    cap = (eval_df[TARGET].to_numpy() == fit_df[TARGET].max())      # rows sitting on the 180-minute cap
    out = {k: [] for k in builders}
    for s in seeds:
        for label, build in builders.items():
            pipe = build(HistGradientBoostingRegressor(random_state=s)).fit(X_of(fit_df), fit_df[TARGET])
            pred = pipe.predict(X_of(eval_df))
            err = np.abs(eval_df[TARGET].to_numpy() - pred)
            out[label].append({**regression_report(eval_df[TARGET], pred),
                               "MAE_on_cap_rows": err[cap].mean(), "MAE_other_rows": err[~cap].mean()})
    return pd.DataFrame({k: pd.DataFrame(v).mean() for k, v in out.items()}).T

print("VALIDATION months (fit on train months), average of 3 seeds")
val_cmp = compare_v1_v2(train, valid)
display(val_cmp.round(3))
print(f"MAE improvement: {val_cmp.iloc[0]['MAE'] - val_cmp.iloc[1]['MAE']:.3f} min "
      f"({100 * (1 - val_cmp.iloc[1]['MAE'] / val_cmp.iloc[0]['MAE']):.1f}% lower)")
val_cmp.round(4).to_csv("reports/v1_vs_v2_validation.csv")
```

```python
import numpy as np
import pickle

# Convert any Pandas StringArray inside fitted transformers to plain NumPy object arrays
def sanitize_for_pickle(obj):
    if hasattr(obj, "named_steps"):
        for step in obj.named_steps.values():
            sanitize_for_pickle(step)
    elif hasattr(obj, "transformers_"):
        for _, trans, _ in obj.transformers_:
            sanitize_for_pickle(trans)
    if hasattr(obj, "categories_"):
        obj.categories_ = [np.asarray(cat, dtype=object) for cat in obj.categories_]
    if hasattr(obj, "feature_names_in_"):
        obj.feature_names_in_ = np.asarray(obj.feature_names_in_, dtype=object)
    if hasattr(obj, "statistics_"):
        obj.statistics_ = np.asarray(obj.statistics_, dtype=object)

sanitize_for_pickle(model)

with open("model_portable.pkl", "wb") as f:
    pickle.dump(model, f)
```

### 17.3 Take-aways

- The model was already close to the best it can do with this data. The remaining error (about 2 minutes) comes mostly from trip-to-trip speed differences that nothing known at dispatch time can explain. Even the diagnostic model that sees the real speed (Section 7.1) scores the same.
- The honest gain from feature engineering here is **small but real**: fewer columns, one strong new column (`total_expected_min`), and a clip at the 180-minute cap.
- Most "clever" features did nothing. Reporting that is part of the result: it shows the features were tested rather than assumed.
- **Not done on purpose:** target-encoding the zones or cuisine (leakage risk, tiny possible gain) and anything that uses `Average_Speed_kmph` (known only after the trip).
- v2 is wired into the whole notebook: model comparison (Section 7), tuning (9), final model (10), intervals (11), interpretation (14), the saved bundle (15) and the independent tests (16). The old `src/features.py` stays in place because v2 builds on it.
